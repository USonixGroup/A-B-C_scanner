```python
from pathlib import Path

import pandas as pd
from IPython.display import display

# file_path = Path("./data/Output voltage calc.xlsx")
file_path = Path("./data/1020L calculation - H102 & H148 transducers at 1st and 3rd harmonics.xlsx")

with pd.ExcelFile(file_path) as workbook:
	for sheet_name in workbook.sheet_names:
		print(sheet_name)
		display(pd.read_excel(workbook, sheet_name=sheet_name))
```

```python
from IPython.display import display, HTML
sheets = pd.read_excel(file_path, sheet_name=None, engine="openpyxl")

for name, df in sheets.items():
    display(HTML(f"<h3>{name}</h3><p>{df.shape[0]} rows × {df.shape[1]} columns</p>"))
    display(
        df.dropna(how="all")
          .dropna(axis=1, how="all")
          .style.set_table_styles(
              [{"selector": "th", "props": [("background-color", "#f0f0f0"), ("text-align", "left")]}]
          )
          .format(precision=4, na_rep="")
    )
```
